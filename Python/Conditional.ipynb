# If-Elif-Else Statement

```python
number = int(input("Enter the number : "))
if number > 0 :
    print("Positive")
elif number < 0 :
    print("Negative")
else :
    print("Zero")
```

```python
marks = int(input("Enter the marks: "))
if marks > 79:
    print("Grade A")
elif marks > 59:
    print("Grade B")
elif marks > 39:
    print("Grade C")
else :
    print("Fail")
```
